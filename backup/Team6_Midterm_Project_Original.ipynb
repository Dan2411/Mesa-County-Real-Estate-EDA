## 3. Data Scraping

The Mesa County website displays property sales in an interactive table. We selected non-overlapping sale-date ranges in the browser, then exported the already-loaded HTML rows to local  `mesa_sales_*.html` files.

Using BeautifulSoup, we extract all original columns from the saved HTML files and combine them with Pandas.

**Input:** Place the exported `mesa_sales_*.html` files inside the `data/` folder of the project repository.

### 3.1 Read the saved HTML tables

```python
html_files = sorted(Path('data').glob('mesa_sales_*.html'))
print('HTML files:', [file.name for file in html_files])
```

**Observation**  
The original collection used three non-overlapping date-range HTML exports. Keeping their filenames allows us to trace each transaction back to its source.

### 3.2 Extract the rows and combine the data

```python
parts = []

for file in html_files:
    # Read the HTML file and find the sales table
    soup = BeautifulSoup(file.read_text(encoding='utf-8'), 'html.parser')
    table = soup.find('table')

    # Extract column names and all data rows
    columns = [th.get_text(strip=True) for th in table.select('thead th')]
    rows = [
        [td.get_text(' ', strip=True) for td in tr.select('td')]
        for tr in table.select('tbody tr')
    ]

    # Convert the rows into a DataFrame and track the source file
    df = pd.DataFrame(rows, columns=columns)
    df['SourceFile'] = file.name
    parts.append(df)

# Combine all HTML tables into one dataset
raw = pd.concat(parts, ignore_index=True)

print('Combined dataset:', raw.shape)
display(raw.head(3))
```

**Observation**  
All HTML table rows are combined into one DataFrame, retaining the original 24 website columns. The additional SourceFile column records which export each row came from.

### 3.3 Save the original dataset

```python
raw.to_csv('data/mesa_sales_raw.csv', index=False, encoding='utf-8-sig')
print(f'Rows: {len(raw):,} | Website columns: {len(raw.columns) - 1}')
print('Meets project minimum:', len(raw) >= 15_000 and len(raw.columns) - 1 >= 7)
```

### 5B. Univariate, Bivariate, and Multivariate Analysis

#### 5B.1 Data cleaning and type conversion

```python
# Rename columns for easier analysis
rename = {
    'Parcel Number (Click for map)': 'parcel_id',
    'Account Number (Click for details)': 'account_id',
    'Reception No./Book Page (Click for Doc.)': 'reception_id',
    'Address': 'address', 'Sale Price': 'sale_price',
    'Time Adjusted Sale Price': 'time_adjusted_sale_price',
    'Sale Date': 'sale_date', 'Grantee': 'grantee', 'Grantor': 'grantor',
    'Actual Value': 'assessor_actual_value',
    'Qualification Code': 'qualification_code', 'Neighborhood': 'neighborhood',
    'Economic Area': 'economic_area', 'Property Use': 'property_use',
    'Total Heated Sq Ft': 'heated_sqft', 'Price by Heated SQFT': 'price_per_sqft_reported',
    'Aprox. Acres': 'acres',
    'Architectural Style 1st Building': 'architectural_style',
    'Total # Buildings': 'building_count',
    'Earliest Effective Year Built': 'effective_year_built',
    'Actual Year Built': 'year_built', 'PropUse': 'property_use_code',
    'Bed': 'bedrooms', 'Bath': 'bathrooms', 'SourceFile': 'source_file'
}

# Rename columns and remove extra spaces
work = work.rename(columns=rename)
work = work.apply(lambda col: col.str.strip())

# Convert numerical columns
numeric_cols = [
    'sale_price', 'time_adjusted_sale_price', 'assessor_actual_value',
    'heated_sqft', 'price_per_sqft_reported', 'acres', 'building_count',
    'effective_year_built', 'year_built', 'bedrooms', 'bathrooms'
]

work[numeric_cols] = (
    work[numeric_cols]
    .replace(r'[$,]', '', regex=True)
    .apply(pd.to_numeric, errors='coerce')
)

# Convert dates and extract year/month
work['sale_date'] = pd.to_datetime(
    work['sale_date'], format='%m/%d/%Y', errors='coerce'
)

work['sale_year'] = work['sale_date'].dt.year
work['sale_month'] = work['sale_date'].dt.strftime('%Y-%m')

# Display results
print(work[numeric_cols + ['sale_date']].dtypes)
display(work[['sale_price', 'sale_date', 'heated_sqft', 'bedrooms', 'bathrooms']].head())
```

**Observation**  
The data types were successfully converted. Sale prices, square footage, building counts, and construction years are now stored as numerical values, while sale dates are stored in datetime format. Bedrooms and bathrooms are stored as floating-point numbers to accommodate missing values and decimal values. These changes make the dataset easier to work with for statistical calculations, comparisons, and visualizations in the following sections.

#### 5B.2 Handle missing or unusual values and remove potential duplicates

```python
# Handle missing and unusual values
work.replace('', np.nan, inplace=True)

placeholder_bath = work['bathrooms'] >= 100
work.loc[placeholder_bath, 'bathrooms'] = np.nan

# Calculate building age at the time of sale
work['year_built_after_sale'] = work['year_built'] > work['sale_year']
work['building_age_at_sale'] = work['sale_year'] - work['year_built']
work.loc[work['year_built_after_sale'], 'building_age_at_sale'] = np.nan

# Remove potential repeated sales
before = len(work)
work = work.drop_duplicates(subset=['parcel_id', 'sale_date', 'sale_price']).copy()
duplicates_removed = before - len(work)

# Remove nonpositive sale prices
before = len(work)
work = work[work['sale_price'] > 0].copy()
invalid_prices_removed = before - len(work)

# Flag unusually low sale prices for later investigation
work['low_sale_price_under_10000'] = work['sale_price'] < 10000

# Display cleaning results
print('Bathroom placeholders:', placeholder_bath.sum())
print('Construction year after sale:', work['year_built_after_sale'].sum())
print('Duplicates removed:', duplicates_removed)
print('Nonpositive prices removed:', invalid_prices_removed)
print('Sales under $10,000:', work['low_sale_price_under_10000'].sum())
print('Remaining records:', len(work))
```

**Observation**  
The cleaning process handled missing and unusual values, including replacing 83 invalid bathroom counts with NaN and marking 321 invalid building ages as missing. We also removed 9 potentially repeated sales and 4 records with nonpositive sale prices. The remaining 200 sales below $10,000 were flagged because their unusually low prices could affect the analysis. These records were kept in the cleaned dataset but excluded from selected analyses later in the notebook. After cleaning, the dataset contains 15,342 records, which still meets the project's minimum requirement of 15,000 rows.

#### 5B.3 Univariate Analysis: Descriptive Statistics and Skewness

```python
univariate_cols=['sale_price','time_adjusted_sale_price','assessor_actual_value','heated_sqft','price_per_sqft_reported','acres','bedrooms','bathrooms','building_age_at_sale']

# Calculate descriptive statistics, skewness, and missing values
descriptive_stats=work[univariate_cols].describe().T
descriptive_stats['median']=work[univariate_cols].median()
descriptive_stats['skewness']=work[univariate_cols].skew()
descriptive_stats['missing_values']=work[univariate_cols].isna().sum()
display(descriptive_stats.round(2))

# Classify each variable's distribution based on its skewness
print("Skewness interpretation:")
for col in univariate_cols:
    skew=work[col].skew()
    if skew>0.5:
        interpretation="Positively skewed (right-skewed)"
    elif skew<-0.5:
        interpretation="Negatively skewed (left-skewed)"
    else:
        interpretation="Approximately symmetric"
    print(f"{col}: skewness={skew:.2f} — {interpretation}")
```

**Observation**  
The descriptive statistics show that all nine numerical variables are positively skewed, meaning that most observations have lower values while a smaller number have unusually high values.

The average sale price is approximately $425,363, which is higher than the median of $383,950. This suggests that expensive properties are pulling the mean upward. Heated square footage is also positively skewed, with an average of approximately 1,784 square feet.

Acreage has the highest skewness (52.19), followed by the reported price per square foot (21.02). This indicates that a small number of properties have unusually large land areas or high prices per square foot.

Bedrooms, bathrooms, and building age are also positively skewed, although less strongly than acreage. The missing-value counts show that these three variables contain incomplete observations, which should be considered when interpreting the results.

##### 5B.4.2 Correlation Between Sale Price and Property Features

```python
# Compare Pearson and Spearman correlations with sale price
corr_vars = {
    'sale_price': 'Sale price',
    'heated_sqft': 'Heated sq ft',
    'bathrooms': 'Bathrooms',
    'bedrooms': 'Bedrooms',
    'acres': 'Lot size (acres)',
    'building_age_at_sale': 'Building age',
    'sale_year': 'Sale year',
}
cols = list(corr_vars.keys())

corr_with_price = pd.DataFrame({
    'Pearson': biv[cols].corr()['sale_price'],
    'Spearman': biv[cols].corr(method='spearman')['sale_price'],
})
corr_with_price = (
    corr_with_price.drop('sale_price')
    .rename(index=corr_vars)
    .sort_values('Spearman', ascending=False)
)
display(corr_with_price.round(3))
```

**Observations**  
 Square footage is the variable most closely tied to price, with a Spearman correlation of 0.72. Bathrooms (0.49) and bedrooms (0.46) come next, which makes sense because more rooms usually means a bigger house. Lot size however, is mixed: Pearson says the link is weak (0.13), but Spearman says it's moderate (0.52). A few very large parcels pull the Pearson value down, so Spearman is the better measure here. This is because Pearson measures a straight-line relationship, while Spearman uses ranks, so it is less affected by skewed prices and extreme values. Building age is negative (−0.30), meaning older homes tend to sell for less. Sale year barely matters on its own (0.11), because the price gap between a small house and a large one is much bigger than the change in prices from one year to the next.

```python
# Group homes by square footage to compare sale prices
sqft_bins = [0, 1000, 1500, 2000, 2500, 3000, 4000, np.inf]
sqft_labels = ['<1,000', '1,000-1,499', '1,500-1,999', '2,000-2,499',
               '2,500-2,999', '3,000-3,999', '4,000+']
biv['sqft_band'] = pd.cut(biv['heated_sqft'], bins=sqft_bins, labels=sqft_labels, right=False)
sqft_summary = biv.groupby('sqft_band', observed=True).agg(
    sales=('sale_price', 'count'),
    median_sqft=('heated_sqft', 'median'),
    median_price=('sale_price', 'median'),
    median_price_per_sqft=('price_per_sqft', 'median'),
)
display(sqft_summary.round(0))
```

But the price per square foot doesn't stay the same. It's highest for homes between 1,000 and 1,499 sq ft (256) and drops to 212 for the largest homes. In other words, buyers pay less for each extra square foot once a house gets big.

##### 5B.5.2 Architectural type profiles and clustering

This part of the analysis uses all property types from the cleaned data, so non-residential buildings, such as offices, restaurants, and warehouses are included.

##### Step 1: Identify and Group Architectural Styles

```python
arch = work[~work['low_sale_price_under_10000']].copy()

# Extract the primary architectural style from each record
arch['arch_style'] = arch['architectural_style'].str.strip('-').str.split('--').str[0]

arch['arch_style'].value_counts().hist()
```

```python
# Group sales by arch_style
all_styles = arch.groupby('arch_style').agg(
    sales=('sale_price', 'count'),
    most_common_use=('property_use', lambda s: s.mode()[0]),
    median_sqft=('heated_sqft', 'median'),
).sort_values('sales', ascending=False)

# arch_style's with <5 sales are rare
all_styles['rare'] = all_styles['sales'] < 5

print('Unique architectural styles:', len(all_styles))
print('Rare styles (fewer than 5 sales):', all_styles['rare'].sum(),
      '| Sales in rare styles:', all_styles.loc[all_styles['rare'], 'sales'].sum())

with pd.option_context('display.max_rows', None):
    display(all_styles)
```


**How the rare styles were grouped**

34 arch_styles have < 5 sales, which is too few for statistical analysis, so they are combined into broader categories instead of being dropped. So each rare style was assigned by what its name says the building is used for:

* **Industrial/Flex:** office units and shopping units, small warehouse, manufacturing building, meat processor, hangar etc.
* **Retail, Office & Services:** small retail, beauty, laundry, showroom store, small office, bank
* **Auto:** dealerships, car washes, service garage
* **Food & Drink:** bar/tavern, restaurant/bar condo, brew pub, bagel/coffee
* **Other Health Care:** vet clinic, retirement/nursing home
* **Special Use:** school, preschool, post office, hall, recreation
* **Other Residential:** 3-story, 4-level, terra dome, and small apartment buildings.

```python
# rare styles are too small to describe on their own so they are grouped into broader categories
rare_style_category = {
    'SM SHOP/OFFICE': 'INDUSTRIAL/FLEX', 'COMM CONDO SHOP/OFFICE': 'INDUSTRIAL/FLEX',
    'SM WAREHOUSE': 'INDUSTRIAL/FLEX', 'MANUFACT BLDG': 'INDUSTRIAL/FLEX',
    'MEAT PROCESSOR': 'INDUSTRIAL/FLEX', 'COMM CONDO T- HANGAR': 'INDUSTRIAL/FLEX',
    'SMALL RETAIL': 'RETAIL, OFFICE & SERVICES', 'BARBER/BEAUTY': 'RETAIL, OFFICE & SERVICES',
    'COMM CONDO BARBER/BEAUTY': 'RETAIL, OFFICE & SERVICES', 'LAUNDRY FACILTY': 'RETAIL, OFFICE & SERVICES',
    'WAREHOUSE SHOWROOM STORE': 'RETAIL, OFFICE & SERVICES', 'SM OFFICE': 'RETAIL, OFFICE & SERVICES',
    'BANK': 'RETAIL, OFFICE & SERVICES',
    'DEALERSHIP': 'AUTO', 'SMALL DLRSHIP': 'AUTO', 'CARWASH': 'AUTO',
    'AUTOMATIC CAR WASH': 'AUTO', 'SERV/GARAGE': 'AUTO',
    'BAR/TAVERN': 'FOOD & DRINK', 'COMM CONDO REST/BAR': 'FOOD & DRINK',
    'BREW PUB': 'FOOD & DRINK', 'BAGEL/COFFEE': 'FOOD & DRINK',
    'VET CLINIC': 'OTHER HEALTH CARE', 'RETIRE/NURSING': 'OTHER HEALTH CARE',
    'SCHOOL': 'SPECIAL USE', 'PRESCHOOL': 'SPECIAL USE', 'POST OFFICE': 'SPECIAL USE',
    'HALL': 'SPECIAL USE', 'RECREATION': 'SPECIAL USE',
    '3 STORY': 'OTHER RESIDENTIAL', '4 LEVEL': 'OTHER RESIDENTIAL',
    'TERRA DOME': 'OTHER RESIDENTIAL', 'APT/MINIMAL': 'OTHER RESIDENTIAL',
    'MULTI 4-8 C': 'OTHER RESIDENTIAL',
}

def broad_category(style):
    return rare_style_category.get(style, 'OTHER')

style_counts = arch['arch_style'].value_counts()
rare_styles = style_counts[style_counts < 5].index
is_rare = arch['arch_style'].isin(rare_styles)
print('Rare styles (fewer than 5 sales):', len(rare_styles), '| Sales:', is_rare.sum())
arch.loc[is_rare, 'arch_style'] = arch.loc[is_rare, 'arch_style'].apply(broad_category)

# Show which rare styles went into which category
rare_map = pd.DataFrame({'style': rare_styles, 'category': [broad_category(x) for x in rare_styles]})
for category, styles in rare_map.groupby('category')['style']:
    print(f'{category}: {", ".join(styles)}')

# every rare style should have a category
print('Rare styles without a category:', (rare_map['category'] == 'OTHER').sum())

# rare styles next to property use & catgory
rare_table = all_styles[all_styles['rare']].copy()
rare_table['category'] = rare_table.index.map(broad_category)
display(rare_table.drop(columns='rare').sort_values(['category', 'sales'], ascending=[True, False]))

arch['price_per_sqft'] = arch['sale_price'] / arch['heated_sqft']
arch['is_home'] = arch['property_use'].isin(['Residential', 'Townhouse', 'Condo'])

style_profile = arch.groupby('arch_style').agg(
    sales=('sale_price', 'count'),
    median_price=('sale_price', 'median'),
    median_sqft=('heated_sqft', 'median'),
    median_price_per_sqft=('price_per_sqft', 'median'),
    median_age=('building_age_at_sale', 'median'),
    pct_home_sales=('is_home', 'mean'),
)
style_profile['pct_home_sales'] = style_profile['pct_home_sales'] * 100

print('Styles and grouped categories:', len(style_profile))
print('Sales covered:', style_profile['sales'].sum(), 'of', len(arch))
print('Smallest group size:', style_profile['sales'].min())
```

##### Step 2: Prepare Features and Determine the Number of Clusters

```python
# Cluster styles by price, size, and building age
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Price and size are skewed, so we use a log scale
features = pd.DataFrame({
    'log_price': np.log10(style_profile['median_price']),
    'log_sqft': np.log10(style_profile['median_sqft']),
    'median_age': style_profile['median_age'],
})

# Put the features on the same scale
features_scaled = StandardScaler().fit_transform(features)

# Use the elbow method to compare cluster counts from 1 to 8
k_list = []
for k in range(1, 9):
    km = KMeans(n_clusters=k, random_state=1, n_init=10)
    km.fit(features_scaled)
    k_list.append(km.inertia_)

plt.figure(figsize=(7, 4))
plt.plot(range(1, 9), k_list, marker='o')
plt.title('Elbow Method for Clustering Architectural Styles')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia (within-cluster error)')
plt.tight_layout()
plt.show()
```

##### Step 3: Cluster Profiles and Compare Results

```python
# Fitting w/ 4 clusters
style_kmeans = KMeans(n_clusters=4, random_state=1, n_init=10)
style_profile['cluster'] = style_kmeans.fit_predict(features_scaled)

# Rank clusters by median style-level sale price
group_rank = style_profile.groupby('cluster')['median_price'].median().rank().astype(int)
style_profile['style_group'] = 'Group ' + style_profile['cluster'].map(group_rank).astype(str)

# Group summaries
group_summary = style_profile.groupby('style_group').agg(
    styles=('sales', 'count'),
    total_sales=('sales', 'sum'),
    median_price=('median_price', 'median'),
    median_sqft=('median_sqft', 'median'),
    median_price_per_sqft=('median_price_per_sqft', 'median'),
    median_age=('median_age', 'median'),
    pct_home_sales=('pct_home_sales', 'mean'),
)
display(group_summary.round(0))

display(style_profile.sort_values(['style_group', 'median_price'])
        .drop(columns='cluster').round(0))
```

**Observation**  
The dataset contains 75 architectural styles, with 34 styles having fewer than five sales (72 sales in total). Instead of removing these rare styles, we grouped them into seven broader categories based on their building uses. #### Question 5. Is there a seasonal pattern in sale prices, and how large is it?

The monthly chart suggested that prices are higher in spring and summer. However, prices also rose over time, so a simple monthly average mixes the season with the long-term trend. To remove the trend, we divide each sale price by the median price of its sale year. A value of 1.03 means the sale was 3% above that year's median. We then compare these relative prices across calendar months and test whether peak-season (April–August) and off-season (November–February) prices differ.

```python
biv['price_vs_year_median'] = biv['sale_price'] / biv.groupby('sale_year')['sale_price'].transform('median')
biv['sale_month_num'] = biv['sale_date'].dt.month

month_effect = biv.groupby('sale_month_num').agg(
    sales=('sale_price', 'count'),
    relative_price=('price_vs_year_median', 'median'),
)
month_effect.index = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
                      'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
display(month_effect.T.round(3))

peak = biv.loc[biv['sale_month_num'].isin([4, 5, 6, 7, 8]), 'price_vs_year_median']
off = biv.loc[biv['sale_month_num'].isin([11, 12, 1, 2]), 'price_vs_year_median']
u_stat, p_value = stats.mannwhitneyu(peak, off)
print(f'Mann-Whitney U test p-value: {p_value:.4g}')
print(f'Peak season median: {peak.median():.3f} | Off-season median: {off.median():.3f}')

fig, ax = plt.subplots(figsize=(9, 4.5))
offsets = (month_effect['relative_price'] - 1) * 100
colors = ['#2a78d6' if v >= 0 else '#eb6834' for v in offsets]
ax.bar(month_effect.index, offsets, color=colors, width=0.6)
for x, v in enumerate(offsets):
    ax.annotate(f'{v:+.1f}%', (x, v), textcoords='offset points',
                xytext=(0, 4 if v >= 0 else -12), ha='center', fontsize=9)
ax.axhline(0, color='black', linewidth=0.8)
ax.set_ylim(offsets.min() - 1.5, offsets.max() + 1.2)
ax.set_ylabel('Median price vs. yearly median (%)')
ax.set_xlabel('Sale month')
ax.set_title('Seasonal price effect after removing the yearly trend')
ax.spines[['top', 'right']].set_visible(False)
ax.grid(axis='y', alpha=0.3)
ax.set_axisbelow(True)
plt.tight_layout()
plt.show()
```

**Observation**  
The results suggest a seasonal pattern in residential sale prices. After adjusting each sale price relative to its year's median, May and June have the highest median relative prices, approximately 4.6% above the yearly median. January and February are about 6.4% and 5.5% below the yearly median, respectively.

Overall, April–August sales have a median relative price 2.5% above the yearly median, while November–February sales are 3.5% below it, a difference of approximately 6 percentage points. The Mann–Whitney U test also indicates a statistically significant difference between these periods.

However, the pattern is not consistent across every month, and differences in the properties sold may also contribute to the results. Therefore, the findings suggest seasonality in the observed sale prices but do not establish that buying in a particular month guarantees a lower price.

#### Question 9. Among homes of similar size, do newer homes sell for more than older homes?


Newer homes are often bigger, so comparing new and old homes directly would have some interference due to the relationship between age and size. To separate them, we group sales by size band and by building age at the time of sale, and compare median prices within each size band.

```python
# Use cleaned residential, townhouse, and condo sales
q9_data = biv[['sale_price', 'heated_sqft', 'building_age_at_sale']].copy()

# Exclude records without a valid building age
missing_age = q9_data['building_age_at_sale'].isna().sum()
q9_data = q9_data.dropna(subset=['building_age_at_sale'])
q9_data = q9_data[q9_data['heated_sqft'] > 0].copy()

# Group homes by age and heated square footage
q9_data['age_group'] = pd.cut(
    q9_data['building_age_at_sale'],
    bins=[-1, 10, 30, 50, np.inf],
    labels=['0-10 yrs', '11-30 yrs', '31-50 yrs', '51+ yrs']
)

q9_data['size_band'] = pd.cut(
    q9_data['heated_sqft'],
    bins=[0, 1500, 2000, 2500, np.inf],
    labels=['Under 1,500', '1,500-1,999', '2,000-2,499', '2,500+'],
    right=False
)

# Compare sale counts and median prices across the groups
q9_count = q9_data.pivot_table(
    values='sale_price',
    index='size_band',
    columns='age_group',
    aggfunc='count',
    observed=True
)

q9_price = q9_data.pivot_table(
    values='sale_price',
    index='size_band',
    columns='age_group',
    aggfunc='median',
    observed=True
)

print('Sales excluded due to missing or invalid age:', missing_age)
print('Records analyzed:', len(q9_data))
display(q9_count)
display(q9_price.round(0))

# Compare the newest homes with homes aged 31-50 years
newer_premium = (
    q9_price['0-10 yrs'] / q9_price['31-50 yrs'] - 1
) * 100

print('Newest vs. 31-50 years old (%):')
print(newer_premium.round(1))
```

#### Question 10. How do recorded sale prices compare with the assessor's actual property values, and does the difference vary by residential property type?

The dataset includes both the sale price and the assessor's recorded actual value for each property. We can compare these two values to see whether properties generally sell above or below their recorded assessor values.

To investigate this, we will use the cleaned residential dataset and calculate the ratio between sale price and assessor actual value. A ratio above 1 means the property sold for more than its recorded assessor value, while a ratio below 1 means it sold for less. We will then compare the median ratios across residential homes, townhouses, and condos.

```python
# Use cleaned residential sales with valid assessor values
q10 = biv[
    (biv['sale_price'] > 0) &
    (biv['assessor_actual_value'] > 0)
].copy()

# Calculate the sale-to-assessor ratio for each property
q10['sale_to_assessor_ratio'] = (
    q10['sale_price'] / q10['assessor_actual_value']
)

# Summarize the comparison by property type
q10_summary = q10.groupby('property_use').agg(
    sales=('sale_price', 'size'),
    median_sale_price=('sale_price', 'median'),
    median_assessor_value=('assessor_actual_value', 'median'),
    median_ratio=('sale_to_assessor_ratio', 'median'),
    pct_above_assessor=(
        'sale_to_assessor_ratio',
        lambda s: (s > 1).mean() * 100
    )
)

# Convert the median ratio to a percentage difference
q10_summary['median_gap_pct'] = (
    q10_summary['median_ratio'] - 1
) * 100

q10_summary = q10_summary.sort_values(
    'median_gap_pct', ascending=False
)

print('Records analyzed:', len(q10))
print(
    'Overall median sale-to-assessor ratio:',
    round(q10['sale_to_assessor_ratio'].median(), 3)
)
print(
    'Sales above assessor value:',
    f"{(q10['sale_to_assessor_ratio'] > 1).mean() * 100:.1f}%"
)

display(q10_summary.round({
    'median_sale_price': 0,
    'median_assessor_value': 0,
    'median_ratio': 3,
    'pct_above_assessor': 1,
    'median_gap_pct': 1
}))

# Compare median sale prices and assessor values by property type
plot_data = q10_summary.sort_values('median_sale_price')
y = np.arange(len(plot_data))

plt.figure(figsize=(9, 4))

plt.hlines(y, plot_data['median_assessor_value'],
           plot_data['median_sale_price'], color='lightgray', linewidth=4)

for col, color, label, offset in [
    ('median_assessor_value', 'steelblue', 'Assessor Value', -12),
    ('median_sale_price', 'darkorange', 'Sale Price', 12)
]:
    plt.scatter(plot_data[col], y, color=color, s=120, label=label, zorder=3)
    for i, value in enumerate(plot_data[col]):
        plt.annotate(f'${value/1000:,.0f}K', (value, i),
                     xytext=(offset, 10), textcoords='offset points',
                     ha='center', fontsize=9)

plt.yticks(y, plot_data.index)
plt.ylim(-0.4, len(plot_data) - 0.2)
plt.xlabel('Median Value ($)')
plt.title('Median Sale Price vs. Assessor Value by Property Type')
plt.xlim(plot_data['median_assessor_value'].min() * 0.95,
         plot_data['median_sale_price'].max() * 1.08)
plt.legend()
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

**Observation**  
The analysis includes 14,133 residential property sales with valid sale prices and assessor actual values. Overall, the median sale-to-assessor ratio is approximately 1.154, meaning that the typical recorded sale price was about 15.4% higher than the corresponding assessor value. Approximately 85.6% of the analyzed properties sold above their recorded assessor values.

When comparing property types, townhouses have the largest median percentage difference at 15.7%, followed closely by residential homes at 15.5%. Condos have the smallest difference at approximately 12.9%. This suggests that sale prices are generally higher than recorded assessor values across all three property types, although the size of the difference varies slightly.

Overall, the results show a consistent gap between recorded sale prices and assessor actual values. However, the assessor values may not reflect the same valuation date as each sale, so these differences do not necessarily indicate inaccurate assessments. The analysis describes how the two recorded values compare rather than determining which one represents the property's true market value.
